## 3. Lists and indexing

A **list** stores several values in one variable.

Python indexing starts at 0:

- `force_data[0]` is the first value
- `force_data[1]` is the second value
- `force_data[-1]` is the last value


````python
force_data = [240, 510, 760, 980]

print("first value =", force_data[0])
print("second value =", force_data[1])
print("last value =", force_data[-1])
print("number of values =", len(force_data))
````

You can add a value to the end of a list using `.append()`.

````python
force_data = [240, 510, 760, 980]
force_data.append(1210)

print(force_data)
````

## 4. Arrays, vectors, and matrices

A Python list is useful for storing values, but many engineering calculations use **NumPy arrays**. NumPy arrays make it easy to do vector and matrix operations.

You will see NumPy arrays in later labs when data is loaded from a file, stored in columns, plotted, or used as a state vector.

The standard import is:

```python
import numpy as np
```

````python
import numpy as np # import numpy package as np (alias)

# A list stores values, but math with a list does not behave like vector math.
force_list = [240, 510, 760]
print("force_list * 2 =", force_list * 2)

# A NumPy array supports element-by-element math.
force_array = np.array([240, 510, 760])
print("force_array * 2 =", force_array * 2)
````

Array math often removes the need for a loop. The next cell converts several temperature measurements from Celsius to Kelvin at once.


````python
import numpy as np

temperature_C = np.array([21.5, 22.0, 22.4, 23.1])
temperature_K = temperature_C + 273.15

print("temperature_C =", temperature_C)
print("temperature_K =", temperature_K)
print("number of values =", len(temperature_C))
````

Arrays can also store vectors and matrices. In Lab 2B, the state of the rocket booster is stored as a vector containing angle and angular rate.

The `@` operator performs matrix multiplication.

````python
import numpy as np

x = np.array([3.00, -0.20])       # state vector: [angle, angular_rate]
I = np.identity(2)                # 2 by 2 identity matrix
y = I @ x                         # matrix-vector multiplication

print("x =", x)
print("I =")
print(I)
print("y =", y)
print("shape of x =", x.shape)
print("shape of I =", I.shape)
print("angle =", x[0])
print("angular_rate =", x[1])
````

Many later labs store recorded data in a 2D array. The first index selects rows, and the second index selects columns. A colon `:` means "all rows" or "all columns".

````python
import numpy as np

# Each row is one sample: [time_s, displacement_mm, force_N]
data = np.array([
    [0.00, 0.00, 0.0],
    [0.50, 0.12, 240.0],
    [1.00, 0.25, 510.0],
] )

time = data[:, 0]
displacement = data[:, 1]
force = data[:, 2]

print("data shape =", data.shape)
print("time column =", time)
print("displacement column =", displacement)
print("force column =", force)
print("last force =", force[-1])
````

## 7. Loops

A `for` loop repeats code for each item in a list.


````python
theta_data = [2.60, 2.75, 2.90, 3.02]

for theta in theta_data:
    print("theta =", theta)
````

A loop can calculate a new value for each measurement. Here, the loop calculates the difference between a target angle and each measured angle. This is the kind of repeated calculation you will see again in Lab 2B.


````python
theta_ref = 3.14
theta_data = [2.60, 2.75, 2.90, 3.02]

for theta in theta_data:
    error = theta_ref - theta
    print("theta =", theta, "error =", error)
````

Another method is to use an index loop.

## 8. Debugging with `print()`

When your code go run, how do you fix it. There are many ways to debug your python code, we introduce a simple but powerful tool: `print()`.

When debugging, it is often useful to print:

- the value of an input variable such as `temperature_C`
- the value of a calculated result such as `temperature_K`
- the type of a value using `type(...)`

When you get an error message, start by reading the **last line**. It usually names the type of error.


### Common beginner errors

- `NameError`: a variable name is misspelled or has not been created yet.
- `SyntaxError`: Python cannot understand the code, often because of a missing colon or parenthesis.
- `IndentationError`: code is not indented correctly.
- `TypeError`: an operation is being applied to the wrong type of value.
- `IndexError`: a list index is outside the valid range.
