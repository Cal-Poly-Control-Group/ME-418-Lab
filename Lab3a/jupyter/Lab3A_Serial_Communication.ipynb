This notebook opens the serial port once, sends one command at a time, and reads each response through `END`.
```python
# Serial opens the USB serial connection to the Nucleo board.
from serial import Serial

# list_ports helps you find the port name for your board.
from serial.tools import list_ports
import time
```

List the serial ports and choose the one associated with the Nucleo board. On Windows, it will usually look like `COM3`. On macOS, use the `/dev/cu.usbmodem...` port.

```python
for port in list_ports.comports():
    print(port.device, port.description)
```

Set the serial port name below before running the rest of the notebook.

```python
# TODO 1: Change this value to match your Nucleo serial port.
SERIAL_PORT = "COM3"  # Change this. macOS example: "/dev/cu.usbmodem14303"
BAUDRATE = 115200
```

Open the serial port and restart the MicroPython program. The notebook sends control characters to stop any previous MicroPython state, soft reset the board, and wait until `main.py` prints `READY LAB3A_SERIAL_V1`.

Thonny must be disconnected from the MicroPython interpreter before this cell is run because only one program can use the serial port at a time. If this cell does not print `READY LAB3A_SERIAL_V1`, go back to Thonny and check that your completed file was uploaded to `/flash` with the exact name `main.py`.

```python
ser = Serial(SERIAL_PORT, baudrate=BAUDRATE, timeout=0.2)
time.sleep(0.3)
ser.reset_input_buffer()

# Ctrl-B returns to normal REPL, Ctrl-C stops a running program,
# and Ctrl-D soft resets MicroPython so /flash/main.py runs again.
ser.write(b"\x02")
time.sleep(0.1)
ser.write(b"\x03")
time.sleep(0.1)
ser.write(b"\x04")

deadline = time.time() + 5
ready = False

while time.time() < deadline and ready == False:
    line = ser.readline().decode().strip()
    if line == "READY LAB3A_SERIAL_V1":
        ready = True
        print(line)

if ready == False:
    print("ERR did not receive READY. Check that /flash/main.py is uploaded and has no errors.")
else:
    print("Serial port is open and main.py is running.")
```
