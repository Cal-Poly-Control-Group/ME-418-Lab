## 2. Define `run_command()`

`run_command()` sends one complete command through the open serial connection
and reads the response through `END`. After responding, the MicroPython program
immediately waits for another command.

**Provided code — do not modify this cell.**

```python
# PROVIDED CODE — DO NOT MODIFY THIS CELL.
def run_command(command, timeout=20):
    lines = []

    ser.write((command + "\r\n").encode())

    deadline = time.time() + timeout
    finished = False

    while time.time() < deadline and finished == False:
        current_line = ser.readline().decode().strip()

        if current_line != "":
            lines.append(current_line)

            if current_line == "END":
                finished = True
            elif current_line.startswith("DATA,") == False:
                print(current_line)

    if finished == False:
        print("ERR no END before timeout")

    return lines
```

## 3. Define the Plotting Function

This provided function loads one saved data file and compares the commanded
velocity profile with the measured motor velocity.

**Provided code — do not modify this cell.**

```python
# PROVIDED CODE — DO NOT MODIFY THIS CELL.
def plot_profile(kp, ki, filename):
    data = np.genfromtxt(filename, delimiter=",")

    times = data[:, 0]
    setpoints = data[:, 1]
    velocities = data[:, 2]

    plt.figure(figsize=(7, 4))
    plt.plot(times, setpoints, "--", label="Velocity setpoint")
    plt.plot(times, velocities, label="Measured velocity")
    plt.xlabel("Time [s]")
    plt.ylabel("Angular velocity [rad/s]")
    plt.title(f"Profile Tracking: Kp = {kp}, Ki = {ki}")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

## 4. Set the Profile and Controller Parameters

Enter the steady velocity, steady-velocity time, acceleration/deceleration time,
stop time, controller gains, and a unique filename for this experiment.

```python
# TODO: Enter the profile and controller parameters specified in the lab manual.
OMEGA_SS_RAD_S = 500
STEADY_TIME_S = 2
ACCELERATION_TIME_S = 1
STOP_TIME_S = 1
KP = 0.2
KI = 0.4
DATA_FILENAME = "data1.csv"
```

## 5. Run One Profile-Tracking Experiment

The provided cell constructs a `RUN_PROFILE` command from the parameters
above. Do not edit the command phrase or communication code.

```python
# PROVIDED COMMAND CODE — DO NOT MODIFY THIS CELL.
response_lines = run_command(
    (
        f"RUN_PROFILE {OMEGA_SS_RAD_S} {STEADY_TIME_S} "
        f"{ACCELERATION_TIME_S} {STOP_TIME_S} {KP} {KI}"
    ),
    timeout=(
        2 * ACCELERATION_TIME_S
        + STEADY_TIME_S
        + STOP_TIME_S
        + 10
    ),
)
```

## 6. Save the Returned Data

Extract the `DATA` lines from the most recent response and save their time,
setpoint, and measured-velocity columns to the selected CSV file.

**Provided code — do not modify this cell.**

```python
# PROVIDED CODE — DO NOT MODIFY THIS CELL.
data_lines = [
    line[len("DATA,"):]
    for line in response_lines
    if line.startswith("DATA,")
]

if len(data_lines) == 0:
    print("ERR no data were returned by the microcontroller")
else:
    with open(DATA_FILENAME, "w") as data_file:
        for data_line in data_lines:
            data_file.write(data_line + "\n")

    print(f"Saved {len(data_lines)} samples to {DATA_FILENAME}.")
```

## 7. Plot the Results

Plot the most recently saved profile-tracking experiment.

**Provided code — do not modify this cell.**

```python
# PROVIDED CODE — DO NOT MODIFY THIS CELL.
plot_profile(KP, KI, DATA_FILENAME)
```
